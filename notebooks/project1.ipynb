# **1.Trim whitespace and standardize order, customer, and transaction IDs; map aliases to master IDs**

Remove extra spaces from column names

```python
df.columns = df.columns.str.strip()
```

# 18. Separate cancelled, test, and fraud-review orders from fulfilled revenue using explicit rules.

Convert status values to lowercase

```python
cancelled = df[df["Status"].str.contains("cancel", na=False)]
```

```python
test_orders = df[
    df["Status"].str.contains("test", na=False) |
    df["Event_Type"].str.contains("test", na=False)]
```
